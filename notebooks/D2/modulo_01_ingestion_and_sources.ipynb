# Dia 02 - Ingestão de dados de fontes heterogêneas
- Apresentar a coleta de dados a partir de diferentes origens e entender os desafios de ingestão. (requisições de API e parsing de formatos)

## Sessões
1. Ingestão de arquivos csv e json
    - Leitura de arquivos com dados brutos
    - Leitura de dados estruturados (csv, json) utilizando pandas ou outra ferramenta equivalente
2. Ingestão via API REST
    - Requisição http 
    - Manipulação de paginação e tratamento de erros de conexão
3. Ingestão de banco relacional (oltp)
    - Conexão a um banco SQLite local utilizando sqlite3 para simular a extração de uma banco de dados relacional
